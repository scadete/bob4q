# 02 — Magic Coins: Entanglement with IBM Bob

In [Notebook 01](./01-hello-quantum-with-bob.ipynb), you flipped a single quantum coin. Now, what happens if we have **two coins**?

### What is Entanglement?
Imagine having two magic coins. When you flip them:
- Coin A might land on Heads or Tails at random (50/50).
- But **Coin B always lands on the exact same face as Coin A**!
- If Coin A is `0`, Coin B is instantly `0`. If Coin A is `1`, Coin B is instantly `1`.
- You will **never** see `01` or `10`.

This connection is called **Quantum Entanglement** (or a **Bell State**).

## Step 1 — Ask Bob to Create Entanglement
We ask IBM Bob:
> *"Bob, how do I entangle two qubits so their measurements always match?"*

Bob answers:
> *"Use a Hadamard (`h`) gate on qubit 0 to create superposition, and then a CNOT (`cx`) gate to link qubit 0 to qubit 1!"*

Let's test Bob's explanation:


````python
bell_code = """from qiskit import QuantumCircuit

qc = QuantumCircuit(2, 2)
qc.h(0)        # Spin qubit 0
qc.cx(0, 1)    # CNOT gate: entangles qubit 1 with qubit 0
qc.measure([0, 1], [0, 1])
"""

print(explain_with_bob(bell_code))
````

## Step 2 — Build and Draw the Entangled Circuit
Let's assemble the circuit:


````python
qc = QuantumCircuit(2, 2)
qc.h(0)
qc.cx(0, 1)
qc.measure([0, 1], [0, 1])

draw_circuit(qc, title="Entangled Pair (Bell State Circuit)")
````

## Step 3 — Run the Magic Coins
Let's run 1,024 shots and see if the two qubits always match:


````python
counts = run_and_plot(qc, shots=1024, title="Entanglement Results: Only 00 and 11!")
print("Observed outcomes:", counts)

# Verify entanglement
only_matching = set(counts.keys()) <= {"00", "11"}
print(f"\n✨ Perfect correlation verified: {only_matching}!")
````

Notice that `01` and `10` **never** occurred! 
Even though each qubit by itself is completely random, together they are 100% synchronized.
