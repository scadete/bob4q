### The Quantum Riddle: The Secret Coin Trick
Imagine someone hides a secret rule inside a black box (an **Oracle**):
- **Secret = 0 (No Phase Flip):** Leaves the spinning coin untouched.
- **Secret = 1 (Phase Flip):** Flips the hidden quantum phase (`+` to `-`) while the coin is spinning in superposition.

By wrapping the secret box between two Hadamard (`H`) gates (`H → Secret Phase Oracle → H`), **quantum phase kickback and interference** convert that invisible phase flip into a guaranteed `0` or `1` measurement in a **single quantum query (`shots=1`)**!

## Step 1 — Bob Builds the Riddle Circuit
Let's set a secret bit (`SECRET = 1`) and ask Bob to build the quantum circuit that detects it:


````python
SECRET = 1  # Try changing this to 0 or 1!

qc = QuantumCircuit(1, 1)

# 1. Put the qubit into superposition
qc.h(0)
qc.barrier()

# 2. Secret Oracle: if SECRET == 1, flip the quantum phase with a Z gate
if SECRET == 1:
    qc.z(0)
qc.barrier()

# 3. Interfere the phases back into a definite 0 or 1 outcome
qc.h(0)

# 4. Measure in a single shot to reveal the secret
qc.measure(0, 0)

draw_circuit(qc, title=f"Quantum Riddle Circuit (Secret = {SECRET})")
````

## Step 2 — Run the Riddle Solver (Single Query!)
Let's run the circuit with **just 1 shot (`shots=1`)** on the simulator and see if quantum interference identifies the secret immediately:


````python
counts = run_circuit(qc, shots=1)
detected = list(counts.keys())[0]

print(f"🕵️ Secret chosen: {SECRET}")
print(f"✨ Single-shot quantum measurement detected: {detected} (counts={counts})")
assert int(detected) == SECRET, "Riddle solved!"
print("🎉 Riddle solved in 1 shot with 100% accuracy!")
````
