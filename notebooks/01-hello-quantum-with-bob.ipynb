### What is a Qubit?
- **Classical bit (regular computer):** Like a standard light switch. It is either completely **OFF (0)** or completely **ON (1)**.
- **Quantum bit / Qubit (quantum computer):** Like a **spinning coin**. While it spins in mid-air, it is not just heads or tails—it is a **superposition** of both! When you catch it (measure it), it lands on 0 or 1.

## Step 2 — Ask Bob to Create a Quantum Coin Toss
Imagine you are chatting with IBM Bob and say:

> *"Hey Bob, write a simple Qiskit circuit that flips a quantum coin with 50/50 odds."*

Let's see what code Bob generates and ask Bob to explain each line:


````python
coin_toss_code = """from qiskit import QuantumCircuit

# Create 1 qubit and 1 classical bit to record the outcome
qc = QuantumCircuit(1, 1)

# Apply Hadamard gate (H) to spin the coin into superposition
qc.h(0)

# Catch the coin and read the result
qc.measure(0, 0)
"""

# Let's ask Bob for an explanation:
print(explain_with_bob(coin_toss_code))
````

## Step 3 — Draw the Circuit
Let's build the circuit in Python and visualize it as a diagram.


````python
from qiskit import QuantumCircuit

# 1. Build the circuit
qc = QuantumCircuit(1, 1)
qc.h(0)          # H gate spins qubit 0
qc.measure(0, 0) # Measure qubit 0 into classical bit 0

# 2. Draw the diagram
draw_circuit(qc, title="Your First Quantum Circuit (Coin Toss)")
````

## Step 4 — Run It on the Simulator
Now we run the circuit 1,024 times ("shots"). 
Because the coin is in equal superposition, we expect approximately **50% 0** and **50% 1**.


````python
# Run and plot the results
counts = run_and_plot(qc, shots=1024, title="Quantum Coin Toss Results (1024 shots)")
print("Final Counts:", counts)
````
