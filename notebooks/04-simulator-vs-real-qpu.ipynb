## Step 2 — Understanding "Shots"
Because quantum physics is probabilistic, running a circuit just once only gives 1 outcome.
To see the probability distribution, we repeat the experiment across multiple **shots** (e.g., 50 vs 1,024 shots).


````python
# Create a 50/50 superposition circuit
qc = QuantumCircuit(1, 1)
qc.h(0)
qc.measure(0, 0)

# Run with only 20 shots
few_shots = run_circuit(qc, shots=20)
print("20 shots (more random variance):", few_shots)

# Run with 1024 shots
many_shots = run_circuit(qc, shots=1024)
print("1024 shots (converges toward 50% / 50%):", many_shots)

plot_counts(many_shots, title="1024 Shots on Local Simulator")
````
